Setup — run this first.

```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

df = pd.read_csv("students.csv")
# quick clean so plots are tidy
df["gender"] = df["gender"].str.strip().str.title()
df["city"]   = df["city"].str.strip().str.title()
for col in ["study_hours", "attendance", "prev_gpa"]:
    df[col] = df[col].fillna(df[col].median())
df = df.drop_duplicates()
df.head()
```

---
## Part 1 — Matplotlib fundamentals

**1.1** Create the arrays `x = np.arange(1, 11)` and `y = x * 3`. Draw a line plot of `x` against `y` with a title, an x-label and a y-label.

```python
# Solution 1.1
x = np.arange(1, 11)
y = x * 3
plt.plot(x, y, marker="o")
plt.title("y = 3x")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

**1.2** Plot a histogram of `final_marks` straight from Pandas (`.plot(kind="hist")`) with 15 bins. Give it a title and an x-label.

```python
# Solution 1.2
df["final_marks"].plot(kind="hist", bins=15, edgecolor="black", title="Distribution of Final Marks")
plt.xlabel("final marks")
plt.show()
```

---
## Part 2 — Distributions, with statistics made visible

**2.1** Plot a histogram of `study_hours`. On top of it, draw two vertical lines: the **mean** in red and the **median** in green (use `plt.axvline`). Add a legend.
Then answer: which line is further to the right, and what does that tell you?

```python
# Solution 2.1
df["study_hours"].plot(kind="hist", bins=20, edgecolor="black")
plt.axvline(df["study_hours"].mean(),   color="red",   linestyle="--", label="mean")
plt.axvline(df["study_hours"].median(), color="green", linestyle="--", label="median")
plt.title("Study Hours — mean vs median")
plt.legend()
plt.show()
# OBSERVATION: the mean is to the RIGHT of the median. The distribution is right-skewed —
# a few students study very long hours and pull the mean upward. The median is the more
# typical value.
```

**2.2** Use Seaborn's `histplot` to plot `final_marks` with a smooth density curve (`kde=True`). Add a title.

```python
# Solution 2.2
sns.histplot(data=df, x="final_marks", kde=True)
plt.title("Final Marks distribution")
plt.show()
```

**2.3** Draw a boxplot of `attendance` using Seaborn. Looking at the plot, are there any outlier dots? Where do they sit — high or low?

```python
# Solution 2.3
sns.boxplot(data=df, y="attendance")
plt.title("Attendance")
plt.show()
# OBSERVATION: any dots below the lower whisker are students with unusually low attendance.
# The box itself shows the middle half (Q1 to Q3) and the line inside is the median.
```

## Part 3 — Categorical plots

**3.1** Draw a `countplot` of `program`. Which program has the most students?

```python
# Solution 3.1
sns.countplot(data=df, x="program")
plt.title("Students per program")
plt.show()
# MOST STUDENTS: read it off the tallest bar (BSCS in this dataset).
```

**3.2** Draw a `barplot` of the average `final_marks` for each `program`.

```python
# Solution 3.2
sns.barplot(data=df, x="program", y="final_marks")
plt.title("Average final marks by program")
plt.ylabel("mean final marks")
plt.show()
```

**3.3** Draw a boxplot of `final_marks` split by `gender` (gender on the x-axis). Which group's marks are more spread out?

```python
# Solution 3.3
sns.boxplot(data=df, x="gender", y="final_marks")
plt.title("Final marks by gender")
plt.show()
# MORE SPREAD: compare the heights of the two boxes — the taller box has the wider
# middle-half spread.
```

## Part 4 — Relationships between variables

**4.1** Draw a scatter plot of `study_hours` (x) against `final_marks` (y). Describe the pattern in one sentence.

```python
# Solution 4.1
sns.scatterplot(data=df, x="study_hours", y="final_marks")
plt.title("Study hours vs final marks")
plt.show()
# PATTERN: as study hours increase, final marks tend to increase — an upward (positive)
# relationship. This is what a correlation of ~0.66 looks like.
```

**4.2** Draw the same relationship as a `regplot` (scatter with a trend line). Does the line slope up or down?

```python
# Solution 4.2
sns.regplot(data=df, x="study_hours", y="final_marks",
            scatter_kws={"alpha": 0.5}, line_kws={"color": "red"})
plt.title("Study hours vs final marks, with trend line")
plt.show()
# SLOPE: the line slopes UPWARD — more study hours are associated with higher marks.
```

**4.3** Draw a scatter of `attendance` (x) vs `final_marks` (y). Compare it visually to the study-hours scatter — which relationship looks stronger (tighter, clearer)?

```python
# Solution 4.3
sns.scatterplot(data=df, x="attendance", y="final_marks")
plt.title("Attendance vs final marks")
plt.show()
# WHICH IS STRONGER: study hours vs marks looks tighter/clearer than attendance vs marks.
# This matches the correlations from last week (study ~0.66 vs attendance ~0.30).
```

**4.4** Draw a `pairplot` of the four numeric columns: `study_hours`, `attendance`, `prev_gpa`, `final_marks`.
On the diagonal you'll see each column's own distribution; off the diagonal, the scatter of each pair.

```python
# Solution 4.4
sns.pairplot(df[["study_hours", "attendance", "prev_gpa", "final_marks"]])
plt.show()
```

## Part 5 — Heatmap

**5.1** Build the correlation matrix of the four numeric columns, then draw it as a heatmap with `annot=True` and the `coolwarm` colour map. Set `vmin=-1, vmax=1` so the colour scale is honest.

```python
# Solution 5.1
corr = df[["study_hours", "attendance", "prev_gpa", "final_marks"]].corr()
sns.heatmap(corr, annot=True, cmap="coolwarm", vmin=-1, vmax=1)
plt.title("Correlation heatmap")
plt.show()
```

**5.2** From the heatmap, which variable has the **strongest** relationship with `final_marks`? Which has the weakest?

**Solution 5.2**

**STRONGEST:** `study_hours` — its cell with `final_marks` is the warmest (highest value, ~0.66).
**WEAKEST:** `prev_gpa` — the palest of the three, with the lowest correlation (~0.28).

**5.3** Print the skewness of `study_hours`. Is it positive or negative, and does that match the shape of its histogram from Part 2?

```python
# Solution 5.3
print("skew:", round(df["study_hours"].skew(), 2))
# MATCHES HISTOGRAM?: yes — the skew is positive (~1.5), and the histogram had a tail
# stretching to the right. Positive skew and a right tail are the same fact.
```
