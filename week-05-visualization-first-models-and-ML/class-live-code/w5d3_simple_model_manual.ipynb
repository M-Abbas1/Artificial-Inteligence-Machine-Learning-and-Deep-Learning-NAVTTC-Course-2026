```python
def cost(y_hat, y):
    c = np.sum((y_hat - y)**2)/2*len(y)
    return c
```

```python
def show(X, y, y_hat):
    plt.scatter(X, y)
    plt.plot(X, y_hat)
    plt.title("Size vs Price")
    plt.xlabel("Size")
    plt.ylabel("Hourse Price")
    plt.show()
```
