### Partition_Split Class
- A simple class that will partition the dataset into dependent and independent variables.
- And split the data into x_train, y_train, x_test, y_test.

```python
class Partition_Split:

    def __init__(self, dataset):
        self.dataset = dataset;
        self.X, self.Y = self.dataset.iloc[:, :-1].values, self.dataset.iloc[:, -1].values;

    def dep_ind(self):
        return self.X, self.Y;
        
    def test_train_split(self, test_size):
        size = int(test_size/1*len(self.X));
        indices = np.arange(len(self.X));
        np.random.shuffle(indices);
        x_test = self.X[indices[:size]];
        x_train = self.X[indices[size:]];
        y_test = self.Y[indices[:size]];
        y_train = self.Y[indices[size:]];
        return x_train, y_train, x_test, y_test;
```

```python
part = Partition_Split(dataset);
```

```python
X, Y = part.dep_ind();
```

```python
x_train, y_train, x_test, y_test = part.test_train_split(1/3);
```

### Implementation of Simple Linear Regression Class

```python
class SimpleLinearRegression:

    def __init__(self):
        self.m = 0.0;
        self.b = 0.0;

    def fit(self, x_train, y_train):
        x_mean = np.mean(x_train);
        y_mean = np.mean(y_train);
        self.m = np.sum([(x_train[i]-x_mean)*(y_train[i]-y_mean) for i in range(len(x_train))])/np.sum([(x-x_mean)**2  for x in x_train]);
        self.b = y_mean-(self.m*x_mean);

    def predict(self, x_test):
        x_test = np.array(x_test);
        return (self.m * x_test) + self.b
```

```python
regressor = SimpleLinearRegression();
```

```python
regressor.fit(x_train, y_train);
```

```python
y_predict = regressor.predict(x_test);
```
