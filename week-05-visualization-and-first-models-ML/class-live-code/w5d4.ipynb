```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
x = np.array([1, 2, 3, 4])
y = np.array([300, 420, 500, 620])
```

```python
plt.scatter(x, y, marker="x")
plt.xlabel("Size")
plt.ylabel("price")
plt.show()
```

```python
m, b = 0, 0
```

```python
def model(x, m, b):
    y_hat = m * x + b
    return y_hat
```

```python
y_hat = model(x, m, b)
```

```python
print(y_hat)
```

```python
print(y)
```

```python
def show(x, y, y_hat):
    plt.scatter(x, y, marker="x")
    plt.plot(x, y_hat)
    plt.xlabel("Size")
    plt.ylabel("price")
    plt.show()
```

```python
def cost(y, y_hat):
    c = np.sum((y_hat - y)**2)/(2*len(y))
    return c
```

```python
cost(y, y_hat)
```
