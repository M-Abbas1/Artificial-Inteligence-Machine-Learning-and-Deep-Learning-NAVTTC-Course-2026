```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
x = np.array([1, 2, 3, 4, 5], dtype=float)      # hours studied
y = np.array([3, 5, 6.5, 9, 10.5])              # marks obtained

for xi, yi in zip(x, y):
    print(f"  studied {xi:.0f} hour(s)  ->  scored {yi}")
```

```python
plt.scatter(x, y, s=100, zorder=3)
plt.xlabel("Hours Studied")
plt.title("Data")
plt.ylabel("Exam Score")
plt.grid()
plt.show()
```

## model 

```python
def predict(x, m, b):
    y_hat = m * x + b
    return y_hat
```

## cost function

```python
def cost(y, y_hat):
    c = (1/(2*len(y))) * np.sum((y_hat - y)**2)
    return c
```

## gradient Descent

```python
def gradients(x, y, y_hat):
    gm = np.sum((y_hat - y)*x)/len(x)
    gb = np.sum(y_hat - y)/len(x)
    return gm, gb
```

```python
def show(x, y, m, b):
    y_hat = predict(x, m, b)
    plt.scatter(x, y, s=60, marker='x', label="Actual Data")
    plt.plot(x, y_hat, label=f"line for m:{m}, b:{b}")
    plt.xlabel("hours studied")
    plt.ylabel("score")
    plt.grid()
    plt.legend()
    plt.show()
```

```python
m, b = 0, 0


y_hat = predict(x, m, b)
print("Actual y : ", y)
print("model prediction : ", y_hat)

loss = cost(y, y_hat)
print(f"when m = {m} and b = {b} the cost = {loss}")
```

```python
show(x, y, m, b)
```

```python
alpha = 0.1
```

```python
## Epoch 1

y_hat = predict(x, m, b)
print(y_hat)

loss = cost(y, y_hat)
print("cost before updating m and b :  ", loss)

gm, gb = gradients(x, y, y_hat)

m = m - alpha * gm
b = b - alpha * gb
```

```python
print(m)
print(b)
```

```python
## Epoch 2

y_hat = predict(x, m, b)
print(y_hat)

loss = cost(y, y_hat)
print("cost before updating m and b :  ", loss)

gm, gb = gradients(x, y, y_hat)

m = m - alpha * gm
b = b - alpha * gb

print("new m : ", m)
print("new b : ", b)
```

```python
show(x, y, m, b)
```
