```python
def calculator():
    num1 = int(input("Enter Frist Number : "))
    num2 = int(input("Enter Second Number : "))
    op = input("Enter +, -, *, / : ")

    result = 0
    if op == '+':
        result = num1 + num2
    elif op == '-':
        result = num1 - num2
    elif op == '*':
        result = num1 * num2
    elif op == '/':
        if num2 == 0:
            print(f"{num1}/{num2},, Division by zero is not possible")
        else:
            result = num1 / num2
    else:
        print("Please Enter a valid operator, (+, -, *, /) supported")

    print(f"Result : {result}")
```
