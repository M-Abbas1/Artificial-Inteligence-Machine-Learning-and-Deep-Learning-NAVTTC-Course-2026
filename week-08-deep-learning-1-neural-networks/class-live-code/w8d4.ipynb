```python
import tensorflow as tf
from tensorflow import keras
```

```python
from tensorflow.keras import Sequential
```

```python
import numpy as np
```

```python
l1 = keras.layers.Dense(units=4, activation='relu')
```

```python
x = np.array([[1, 0]])
```

```python
a1 = l1(x)
a1
```

```python
l2 = keras.layers.Dense(units=1, activation="sigmoid")
```

```python
l2(a1)
```

```python
model = keras.Sequential([l1, l2])
```

```python
model(x)
```

```python
m = keras.Sequential(
    [
        keras.layers.Input((2,)),
        keras.layers.Dense(units=4, activation='relu', name="Layer1"),
        keras.layers.Dense(units=1, activation='sigmoid', name="output layer")]
)
```

```python
m(x)
```

```python
m.summary()
```
