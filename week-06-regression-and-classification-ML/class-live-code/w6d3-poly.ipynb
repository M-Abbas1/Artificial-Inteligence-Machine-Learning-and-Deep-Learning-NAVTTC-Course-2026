```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
```

```python
x = np.arange(1, 51).reshape(-1, 1)

y = x**2
```

```python
plt.scatter(x, y)
```

```python
model = LinearRegression()

model.fit(x, y)
```

```python
pred = model.predict(x)
```

```python
plt.scatter(x, y, color='red', s=40, zorder=3, label="original data")
plt.plot(x, pred, lw=2, c='pink', label="best line")
plt.xlabel("X")
plt.ylabel("y")
plt.grid()
plt.legend()
plt.show()
```

```python
X = np.c_[x, x**2]
```

```python
X
```

```python
print(X.shape)
print(y.shape)
```

```python
poly_model = LinearRegression()
poly_model.fit(X, y)
```

```python
poly_pred = poly_model.predict(X)
```

```python
plt.scatter(X[:, 0], y, color='red', s=40, zorder=3, label="original data")
plt.plot(X[:, 0], poly_pred, lw=2, c='black', label="best line", zorder=4)
plt.xlabel("X")
plt.ylabel("y")
plt.grid()
plt.legend()
plt.show()
```

```python
from sklearn.metrics import r2_score, mean_squared_error
```

```python
r2_score(poly_pred, y)
```

```python
mean_squared_error(poly_pred, y)
```

```python
poly_model.coef_
```

```python
poly_model.intercept_
```
