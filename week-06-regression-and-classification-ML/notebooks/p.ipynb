```python
import numpy as np
import matplotlib.pyplot as plt
import pandas as pd
```

```python
gen = np.random.default_rng(5)

x = np.linspace(0, 6, 30)
y_true = 10 + 5*x - 1.2*x**2 + 0.1*x**3
y = y_true + gen.normal(0, 1.5, len(x))
```

```python
x.shape
```

```python
y.shape
```

```python
plt.scatter(x, y, color='red', s=40, zorder=3)
plt.xlabel("X")
plt.ylabel("Y")
plt.title("This data clearly curve")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
from sklearn.linear_model import LinearRegression
```

```python
X = x.reshape(-1, 1)

model = LinearRegression().fit(X, y)

plt.scatter(x, y, color='red', s=40, zorder=3)
plt.plot(X, model.predict(X))
plt.xlabel("X")
plt.ylabel("Y")
plt.title("This data clearly curve")
plt.grid(True, alpha=0.3)
plt.show()
```

```python
x2 = x**2
x3 = x**3

X_poly = np.column_stack([x, x2, x3])


model = LinearRegression().fit(X_poly, y)
```

```python
plt.scatter(x, y, color='red', s=40, zorder=3)
plt.plot(x, model.predict(X_poly))
plt.xlabel("X")
plt.ylabel("Y")
plt.title("This data clearly curve")
plt.grid(True, alpha=0.3)
plt.show()
```
