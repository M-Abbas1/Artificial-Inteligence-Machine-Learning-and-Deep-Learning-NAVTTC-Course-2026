```python
cor_username = "admin"
cor_password = "admin123"


max_attempt = 3
attempt = 0

while attempt < 3:

    username = input("Enter Username : ")
    password = input("Enter Password : ")
    # attempt += 1

    if username==cor_username and password==cor_password:
        print("Login Successful!")
        break
    else:
        print("Access Denied!")
        print(f"{username} or {password} is incorrect!")
        print(f"You have {max_attempt - attempt} attempt left")

```
