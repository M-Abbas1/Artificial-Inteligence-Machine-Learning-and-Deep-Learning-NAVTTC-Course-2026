```python
try:
    with open(r"files\calculation_data.csv", 'r') as f:
        reader = csv.reader(f)
        list_sal = list(reader)

    salaries = []
    for sal in list_sal[1:]:
        salaries.append(int(sal[0]))

    salary_dict = {list_sal[0][0] : salaries}


    avg_salary = sum(salary_dict['salary'])/len(salary_dict['salary'])

except FileNotFoundError as e:
    print(e)
except ValueError as e:
    print(e)
    print("Missing data, or noisy data")
except ZeroDivisionError as e:
    print(e)
    print("I think their is no salary available in the file")
except:
    print("Something bad happend")

else:
    print(f"The average salary is : {avg_salary}")
finally:
    print("the finally block alwasys executes")
```

```python
try:
    file = open(r"files\calculation_data.csv", 'r')
    reader = csv.reader(file)
    list_sal = list(reader)

    salaries = []
    for sal in list_sal[1:]:
        salaries.append(int(sal[0]))

    salary_dict = {list_sal[0][0] : salaries}


    avg_salary = sum(salary_dict['salary'])/len(salary_dict['salary'])

except FileNotFoundError as e:
    print(e)
except ValueError as e:
    print(e)
    print("Missing data, or noisy data")
except ZeroDivisionError as e:
    print(e)
    print("I think their is no salary available in the file")
except:
    print("Something bad happend")

else:
    print(f"The average salary is : {avg_salary}")
finally:
    file.close()
```
