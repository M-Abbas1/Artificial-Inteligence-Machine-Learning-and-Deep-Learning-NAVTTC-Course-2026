```python
class LIST(list):
    def __add__(self, other):

        small_list = self
        if len(self) > len(other):
            small_list = other

        result = LIST([])
        for i in range(len(small_list)):
            result.append(self[i] + other[i])
            
        if len(self) > len(other):
            result.extend(self[i+1:])
        else:
            result.extend(other[i+1:])
        
        return result

            
```

```python
class LIST(list):
    def __add__(self, other):
        if len(self) > len(other):
            self, other = other, self

        result = LIST([])
        for i in range(len(self)):
            result.append(self[i] + other[i])
            
        if len(self) > len(other):
            result.extend(self[i+1:])
        else:
            result.extend(other[i+1:])
        
        return result

            
```

```python
class ElectricVehicle:
    def recharge(self):
        print("Car is charging!")
    
    def drive(self):
        print("Car in action after recharging")


class GasolineVehicle:
    def refuel(self):
        print("Refueling!!")

    def drive(self):
        print("Car in action after refueling")


class HybridVehicle(ElectricVehicle, GasolineVehicle):
    pass
```

```python
car.drive()
```

```python
class A:
    def m1(self):
        return 20

class B(A):
    def m1(self):
        return 30
    
    def m2(self):
        return 40


class C(B):
    def m2(self):
        return 20 + super().m1()


obj1 = A()
obj2 = B()
obj3 = C()


print(obj1.m1() + obj2.m1() + obj3.m2() + obj3.m1())
```
