```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.model_selection import train_test_split
from sklearn.datasets import load_iris
from sklearn.tree import DecisionTreeClassifier
from sklearn.tree import plot_tree
```

```python
data = load_iris()
```

```python
data.target
```

```python
data.data
```

```python
data.feature_names
```

```python
X = data.data
y = data.target
```

```python
X.shape
```

```python
y.shape
```

```python
tree = DecisionTreeClassifier(max_depth=3, random_state=0)
```

```python
x_train, x_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

```python
tree.fit(x_train, y_train)
```

```python
tree.predict(x_test)
```

```python
y_test
```

```python
tree.score(x_test, y_test)
```

```python
tree.score(x_train, y_train)
```

```python
tree2 = DecisionTreeClassifier(max_depth=2)
```

```python
tree2.fit(x_train, y_train)
```

```python
tree2.score(x_train, y_train)
```

```python
tree2.score(x_test, y_test)
```

```python
fig = plt.figure(figsize=(10, 10))
plot_tree(tree, feature_names=data.feature_names, class_names=data.target_names, filled=True, rounded=True, fontsize=11)
```

```python
fig = plt.figure(figsize=(10, 10))
plot_tree(tree2, feature_names=data.feature_names, class_names=data.target_names, filled=True, rounded=True, fontsize=11)
```
