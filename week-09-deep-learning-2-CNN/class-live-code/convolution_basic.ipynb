```python
import numpy as np
```

```python
# a tiny 6x6 'image' (just numbers for now)
image = np.array([
    [10, 10, 10, 0, 0, 0],
    [10, 10, 10, 0, 0, 0],
    [10, 10, 10, 0, 0, 0],
    [10, 10, 10, 0, 0, 0],
    [10, 10, 10, 0, 0, 0],
    [10, 10, 10, 0, 0, 0],
], dtype=float)

# a 3x3 kernel that detects a vertical edge (left-to-right change)
kernel = np.array([
    [-1, 0, 1],
    [-1, 0, 1],
    [-1, 0, 1],
], dtype=float)

print("image:\n", image.astype(int))
print("\nkernel (vertical-edge detector):\n", kernel.astype(int))
```

```python
out_img = np.zeros((4, 4))

out_img
```

```python
out_img[0,0] = np.sum(image[0:3, 0:3] * kernel)
```
