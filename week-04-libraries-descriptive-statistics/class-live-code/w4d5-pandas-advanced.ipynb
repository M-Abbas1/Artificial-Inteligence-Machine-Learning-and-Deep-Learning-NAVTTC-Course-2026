```python
def numerical_conversion(s):
    match s:
        case "HP":
            return 1
        case "Asus":
            return 2
        case "Dell":
            return 3
        case "Toshiba":
            return 4
        case "Fujitsu":
            return 5
        case "Acer":
            return 6
        case "MSI":
            return 7
        case "Vero":
            return 8
        case "Apple":
            return 9
        case "Razer":
            return 10
        case "Samsung":
            return 11
        case "Lenovo":
            return 12
```

```python
df["Manufacturer"] = df["Manufacturer"].apply(numerical_conversion)
```
